```python
import numpy as np
# From book
def MeanError(estimates, actual):
    errors = [estimate-actual for estimate in estimates]
    return np.mean(errors)
```

```python
def gammaEstimate(r=1,l=2, mu=0.5, sigma=0.25):
  estimates = []
  for _ in range(100):
    sample = np.random.gamma(shape=r, scale=(1/l), size=100000) 
    var_bar = np.var(sample)
    y_bar = np.mean(sample)
    r_hat = (y_bar**2) / var_bar
    lambda_hat = y_bar / var_bar
    estimates.append((r_hat, lambda_hat))
  return r, l, estimates
r, l, estimates = gammaEstimate()
print(f"error r: {MeanError([x[0] for x in estimates], r)}")
print(f"error lambda: {MeanError([x[1] for x in estimates], l)}")
# 
```
